# 🤖 Integrasi IndoSlang-NLP dengan Scikit-Learn Pipeline

Tutorial ini menunjukkan integrasi `IndoSlangTransformer` ke dalam `sklearn.pipeline.Pipeline`.


```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from indoslang.sklearn import IndoSlangTransformer

X_train = [
    'produknya bgs bgt bangeeet, suka bgt!',
    'pelayanan kureeenggg bgt jelek parah',
    'mantul gan barang cpt sampai',
    'kecewa bgt t0x1c sellernya gak ramah'
]
y_train = [1, 0, 1, 0]

pipeline = Pipeline([
    ('normalizer', IndoSlangTransformer(fix_elongation=True, ambiguity_mode='conservative')),
    ('tfidf', TfidfVectorizer()),
    ('classifier', MultinomialNB())
])

pipeline.fit(X_train, y_train)
print('Model berhasil dilatih dengan IndoSlang-NLP preprocessing!')

X_test = ['barangnya bgs bangeeet']
pred = pipeline.predict(X_test)
print(f'Prediksi untuk "{X_test[0]}":', 'Positif' if pred[0] == 1 else 'Negatif')
```
