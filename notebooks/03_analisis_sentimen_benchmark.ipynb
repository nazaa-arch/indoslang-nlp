# 📊 Analisis Sentimen: Pengaruh Normalisasi IndoSlang-NLP

Membandingkan representasi fitur teks sebelum dan sesudah normalisasi slang/singkatan.


```python
import pandas as pd
from indoslang import IndoNormalizer

reviews = [
    'gw suka bgt sma produk ini, mantul bangeeet',
    'jgn beli disini, nyesel parah kureng bgt',
    'packing rapih, respon cepet, makasih bnyk yaa',
    'b0d0h bgt pelayanannya, gak rekomen blass'
]

df = pd.DataFrame({'ulasan_mentah': reviews})
normalizer = IndoNormalizer()
df['ulasan_bersih'] = normalizer.transform_batch(df['ulasan_mentah'].tolist())
print(df.to_string())
```
