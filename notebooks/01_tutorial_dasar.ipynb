# 🇮🇩 IndoSlang-NLP: Tutorial Dasar Normalisasi Teks

Tutorial ini mendemonstrasikan cara menggunakan `indoslang-nlp` untuk menormalkan teks gaul, singkatan, leetspeak, reduplikasi, dan masking PII.


```python
from indoslang import IndoNormalizer

normalizer = IndoNormalizer()
text = 'gw mager bangeeet, bsk ktmuan yaa di Jaksel!!'
print('Input:', text)
print('Output:', normalizer.normalize(text))
```

## 1. De-elongasi dan Leetspeak
Mengatasi huruf berulang (*elongation*) dan penggantian karakter (*leetspeak*) secara deterministik.


```python
sample = 'b0d0h bgt sih lu, parahhh bangeeettt!'
print(normalizer.normalize(sample))
```

## 2. Pelestarian Sinyal Emosi (*Affective Features*)
Mempertahankan sinyal kapitalisasi, tanda baca seru/tanya, dan emoticon untuk model sentimen.


```python
norm_affect = IndoNormalizer(extract_features=True)
doc = norm_affect.process('PARAHHH BANGEEETTT SIHHH GAK JELAS BGT!!! 😡😡')
print('Teks Bersih:', doc.text)
print('Fitur Afektif:', doc.features)
```

## 3. Mode Jejak Transparan (*Explainability Trace*)
Menelusuri setiap mutasi kata langkah demi langkah.


```python
doc_trace = normalizer.process('gw mager bgt', trace=True)
for step in doc_trace.trace_steps:
    print(f'[{step.stage}] {step.original_token} -> {step.transformed_token} ({step.rule_source})')
```
